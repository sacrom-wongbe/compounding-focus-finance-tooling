```python
# ==============================================================
# Risk-OFF PMO Turn Tester  (standalone Jupyter cell)
# --------------------------------------------------------------
# Window   : BASELINE closes below SMA*(1-BASE_BUFFER) ... until it closes above SMA*(1+BASE_BUFFER)
# PMO      : DecisionPoint PMO on BASELINE = EMA(2/LEN2) of 10*EMA(2/LEN1) of 1-bar ROC%
# First entry: the close the window opens (SPY breaks SMA - buffer), only if not skipped AND PMO > -ZERO_BUFFER.
#            Otherwise no first entry; the window waits for a turn entry.
# Turn entry : inside a window, while flat: PMO has fallen ENTER_DAYS closes in a row ("double negative") AND
#            PMO > -ZERO_BUFFER (buy only while momentum is still around/above zero). Never skipped.
# Exits    : turn exit — PMO has risen EXIT_DAYS closes in a row AND PMO < +ZERO_BUFFER (sell only below zero)
#            upper barrier — PMO closes at/above PMO_UPPER_EXIT
#            window end — sold at the close the day BASELINE turns risk-on; optional hard stop loss (intraday)
# Skip     : applies to the FIRST entry of a window only. Swing lows are used only for this. Swing low = PMO value lowest of PIVOT_K bars either
#            side (below zero), usable once confirmed PIVOT_K bars later. Reference = lowest swing low in the
#            lookback, or the most recent one. Skip the entry if PMO <= reference * ratio:
#              ratio 1.1 = already 10% past the previous low, 1.0 = at it, 0.8 / 0.65 = already near it.
# Fills    : entries and PMO / window-end exits at the close; stop loss on daily open/low
# ==============================================================
import numpy as np
import pandas as pd
import yfinance as yf
import matplotlib.pyplot as plt
from IPython.display import display

# ------------------------- INPUTS -------------------------------
BASELINE       = "SPY"
BASE_SMA       = 200
BASE_BUFFER    = 0.04
TRADE          = {"SPY": -3.0}        # traded inverse (synthetic SPXU); real ETF: {"SPXU": 1}
TRADE_COST     = 0.0                  # annual drag on TRADE (e.g. 0.03)
STOP_LOSS      = None                 # e.g. 0.15; None = off

PMO_LEN1, PMO_LEN2 = 35, 20
ENTER_DAYS_LIST = [2]                 # falling PMO closes in a row to enter
EXIT_DAYS_LIST  = [2, 3]              # rising PMO closes in a row to exit
ZERO_BUFFERS    = [0.0, 0.5, 0.2]          # entries need PMO > -b, turn exits need PMO < +b (0 = strictly above/below 0)
PMO_UPPER_EXIT  = 2.2                 # sell when PMO closes at/above this; None = off

PIVOT_K        = 5
PIVOT_BELOW_ZERO = True
LOOKBACKS      = [252, 378]           # ~1 year / ~1.5 years
SKIPS          = [None, ("lowest", 1.1)]
                 # (reference, ratio): skip if PMO <= reference * ratio; None = never skip

BASE_TPS       = [0.04]         # baselines on the same entries (first ZERO_BUFFER, no skip): fixed TP + window end

SHALLOW_DEPTH, FLASH_DAYS = 0.08, 90  # hindsight window tags (analysis only)
CASH_RATE      = 0.03
START, END     = "1994-01-01", "2026-10-23"
SHOW           = (2, 2, 252, ("lowest", 1.05), 0.3)
                 # (ENTER_DAYS, EXIT_DAYS, LOOKBACK, SKIP, ZERO_BUFFER) — always run, even if not in the grid
TRADING_DAYS   = 252
PLOT_FULL      = True                 # full-history chart
PLOT_WINDOWS   = True                 # one 4-panel chart per risk-off window
WINDOW_PAD     = (60, 40)             # bars shown before / after each window
PLOT_WINDOW_TYPES = None              # e.g. ["bear", "flash"] to only chart those; None = all
# ----------------------------------------------------------------

pct = lambda x: f"{x * 100:g}%"
tickers = sorted({BASELINE, *TRADE})
warmup = pd.Timestamp(START) - pd.Timedelta(days=int(max(BASE_SMA, max(LOOKBACKS)) * 1.6) + 200)
raw = yf.download(tickers, start=warmup, end=END, auto_adjust=True, progress=False)
if not isinstance(raw.columns, pd.MultiIndex):
    raw.columns = pd.MultiIndex.from_product([raw.columns, tickers])
raw = raw[["Open", "High", "Low", "Close"]].dropna(how="any")
px = raw["Close"]
rets = px.pct_change()

def synth(weights, cost=0.0):
    w = pd.Series(weights, dtype=float)
    r = rets[w.index].mul(w, axis=1).clip(lower=-w.abs(), axis=1).sum(axis=1)
    r = (r - cost / TRADING_DAYS).clip(lower=-1.0); r.iloc[0] = 0.0
    return (1 + r).cumprod()

def synth_ohl(weights, close):
    w = pd.Series(weights, dtype=float)
    prev = px[w.index].shift(1)
    move = lambda f: raw[f][w.index] / prev - 1
    o = (move("Open") * w).sum(axis=1)
    hi = sum(wt * move("High" if wt > 0 else "Low")[t] for t, wt in w.items())
    lo = sum(wt * move("Low" if wt > 0 else "High")[t] for t, wt in w.items())
    base = close.shift(1)
    return base * (1 + o).clip(lower=0), base * (1 + hi), base * (1 + lo).clip(lower=0)

trd = synth(TRADE, TRADE_COST)
trd_o, trd_h, trd_l = synth_ohl(TRADE, trd)

b = px[BASELINE]
b_sma = b.rolling(BASE_SMA).mean()
g, cur = np.full(len(b), np.nan), None
for k_, (p, s) in enumerate(zip(b.values, b_sma.values)):
    if np.isnan(s): continue
    if cur is None: cur = p > s
    elif cur and p < s * (1 - BASE_BUFFER): cur = False
    elif not cur and p > s * (1 + BASE_BUFFER): cur = True
    g[k_] = cur
gate = pd.Series(g, index=b.index)

roc = b.pct_change() * 100
pmo = (10 * roc.ewm(alpha=2 / PMO_LEN1, adjust=False).mean()).ewm(alpha=2 / PMO_LEN2, adjust=False).mean()

# swing lows and references (full history, no look-ahead)
Pf = pmo.values
piv_i = np.array([i for i in range(PIVOT_K, len(Pf) - PIVOT_K)
                  if not np.isnan(Pf[i]) and Pf[i] == np.nanmin(Pf[i - PIVOT_K:i + PIVOT_K + 1])
                  and (Pf[i] < 0 or not PIVOT_BELOW_ZERO)], dtype=int)
piv_v, piv_c = Pf[piv_i], piv_i + PIVOT_K

def lowest_low(lookback):
    ref = np.full(len(Pf), np.nan)
    if len(piv_i):
        tt = np.arange(len(Pf))[:, None]
        ok = (piv_c[None, :] <= tt) & (piv_i[None, :] >= tt - lookback)
        vals = np.where(ok, piv_v[None, :], np.inf).min(axis=1)
        ref[np.isfinite(vals)] = vals[np.isfinite(vals)]
    return pd.Series(ref, index=pmo.index)

def recent_low(lookback):
    ref = np.full(len(Pf), np.nan)
    if len(piv_i):
        tt = np.arange(len(Pf))
        j = np.searchsorted(piv_c, tt, side="right") - 1
        ok = j >= 0
        ok[ok] &= piv_i[j[ok]] >= tt[ok] - lookback
        ref[ok] = piv_v[j[ok]]
    return pd.Series(ref, index=pmo.index)

idx = px.index[(px.index >= pd.Timestamp(START)) & gate.notna()]
B, BS = b.loc[idx].values, b_sma.loc[idx].values
T, TO, TL = (x.loc[idx].values for x in (trd, trd_o, trd_l))
TH = trd_h.loc[idx].values
P = pmo.loc[idx].values
REFS = {("lowest", lb): lowest_low(lb).loc[idx].values for lb in LOOKBACKS}
REFS.update({("recent", lb): recent_low(lb).loc[idx].values for lb in LOOKBACKS})
ON = gate.loc[idx].astype(bool).values
N = len(idx)

# PMO direction streaks (close-to-close)
DN, UP = np.zeros(N, int), np.zeros(N, int)
for t in range(1, N):
    DN[t] = DN[t-1] + 1 if P[t] < P[t-1] else 0
    UP[t] = UP[t-1] + 1 if P[t] > P[t-1] else 0

windows, i = [], 1
while i < N:
    if ON[i-1] and not ON[i]:
        j = i
        while j < N and not ON[j]: j += 1
        e = min(j, N - 1)
        depth = (B[i:e+1] / BS[i:e+1] - 1).min()
        kind = "shallow" if depth > -SHALLOW_DEPTH else ("flash" if e - i <= FLASH_DAYS else "bear")
        windows.append(dict(s=i, e=e, open=j >= N, depth=depth, type=kind))
        i = j
    i += 1
if not windows:
    raise ValueError("No risk-off windows in this date range.")
win_of = np.full(N, -1)
for wn, w in enumerate(windows):
    win_of[w["s"]:w["e"] + 1] = wn


def skip_level(lb, skip, t):
    if skip is None: return np.nan
    r = REFS[(skip[0], lb)][t]
    return r * skip[1] if not np.isnan(r) and r < 0 else np.nan


def run(nd, nu=None, lb=None, skip=None, tp=None, zb=0.0):
    """nd = falling days to enter, nu = rising days to exit (None = baseline: no PMO exits), tp = fixed TP,
    zb = zero buffer."""
    cash_d = (1 + CASH_RATE) ** (1 / TRADING_DAYS) - 1
    eq, curve, trades, marks = 1.0, np.ones(N), [], []
    in_pos, st, skips, n_entries, blocked = False, {}, 0, 0, 0

    for t in range(1, N):
        wn = win_of[t]
        exited_today = False
        if in_pos:
            ep, hit = st["ep"], None
            if STOP_LOSS:
                L = ep * (1 - STOP_LOSS)
                if TO[t] <= L: hit = (TO[t], "stop loss")
                elif TL[t] <= L: hit = (L, "stop loss")
            if hit is None and tp is not None:
                L = ep * (1 + tp)
                if TO[t] >= L: hit = (TO[t], "take-profit")
                elif TH[t] >= L: hit = (L, "take-profit")
            if hit is None and nu and UP[t] >= nu and t - st["t0"] >= nu and P[t] < zb:
                hit = (T[t], "PMO turn up")
            if hit is None and nu and PMO_UPPER_EXIT is not None and P[t] >= PMO_UPPER_EXIT:
                hit = (T[t], "PMO upper")
            if hit is None and ON[t]: hit = (T[t], "window end")
            if hit:
                fill, why = hit
                eq *= fill / T[t-1]
                trades.append(dict(window=st["wn"] + 1, kind=st["kind"], entry=idx[st["t0"]], exit=idx[t],
                                   days=t - st["t0"], ret=fill / ep - 1, reason=why))
                marks.append(("exit", t, fill, why))
                in_pos, exited_today = False, True
            else:
                eq *= T[t] / T[t-1]
        else:
            eq *= 1 + cash_d

        if not in_pos and not exited_today and wn >= 0 and not ON[t]:
            kind = None
            if t == windows[wn]["s"]:                              # SPY just broke SMA - buffer
                lvl = skip_level(lb, skip, t)
                if not np.isnan(lvl) and P[t] <= lvl:
                    skips += 1
                    marks.append(("skip", t, T[t], "first"))
                elif P[t] > -zb:
                    kind = "first"
                else:
                    blocked += 1
                    marks.append(("blocked", t, T[t], "first"))
            elif DN[t] == nd and P[t] > -zb:
                kind = "turn"
            if kind:
                in_pos, n_entries = True, n_entries + 1
                st = dict(ep=T[t], t0=t, wn=wn, kind=kind)
                marks.append(("entry", t, T[t], kind))
        curve[t] = eq

    if in_pos:
        trades.append(dict(window=st["wn"] + 1, kind=st["kind"], entry=idx[st["t0"]], exit=idx[-1],
                           days=N - 1 - st["t0"], ret=T[-1] / st["ep"] - 1, reason="still open"))
    curve = pd.Series(curve, index=idx)
    wins = [dict(ret=curve.iloc[w["e"]] / curve.iloc[w["s"] - 1] - 1,
                 traded=any(m[0] == "entry" and win_of[m[1]] == wn for m in marks)) for wn, w in enumerate(windows)]
    return dict(curve=curve, trades=pd.DataFrame(trades), marks=marks, wins=wins, skips=skips, blocked=blocked)


REASONS = ["PMO turn up", "PMO upper", "take-profit", "stop loss", "window end", "still open"]

def summarize(res):
    w = pd.DataFrame(res["wins"]).assign(type=[x["type"] for x in windows])
    tr, c, r = res["trades"], res["curve"], np.array([x["ret"] for x in res["wins"]])
    years = (c.index[-1] - c.index[0]).days / 365.25
    out = {"Windows traded": w["traded"].mean(), "Mean window": r.mean(), "Median window": np.median(r),
           "% windows > 0": (r > 0).mean(), "% windows ≥ 10%": (r >= 0.10).mean(),
           "Worst window": r.min(), "Best window": r.max(),
           "Compounded": np.prod(1 + r) - 1, "Compounded ex-best": np.prod(1 + np.sort(r)[:-1]) - 1}
    for k in ["shallow", "flash", "bear"]:
        sub = w[w.type == k]
        out[f"Mean {k} ({len(sub)})"] = sub["ret"].mean() if len(sub) else np.nan
    has = len(tr) > 0
    out.update({"Trades": len(tr), "First entries skipped": res["skips"], "First entries blocked (PMO<0)": res["blocked"],
                "Trade win rate": (tr["ret"] > 0).mean() if has else np.nan,
                "Avg trade": tr["ret"].mean() if has else np.nan,
                "Avg win": tr.loc[tr.ret > 0, "ret"].mean() if has and (tr.ret > 0).any() else np.nan,
                "Avg loss": tr.loc[tr.ret <= 0, "ret"].mean() if has and (tr.ret <= 0).any() else np.nan,
                "Avg days held": tr["days"].mean() if has else np.nan,
                "Sleeve CAGR": c.iloc[-1] ** (1 / years) - 1, "Sleeve max DD": (c / c.cummax() - 1).min()})
    why = tr["reason"].value_counts(normalize=True) if has else pd.Series(dtype=float)
    out.update({f"Exit: {k}": why.get(k, 0.0) for k in REASONS})
    return out


skip_label = lambda sk: "no skip" if sk is None else f"skip {sk[0]}×{sk[1]:g}"
name = lambda nd, nu, lb, sk, zb: (f"PMO turn | in {nd}↓ out {nu}↑ | zero ±{zb:g} | " +
                                   (skip_label(sk) if sk is None else f"lb {lb} | {skip_label(sk)}"))
results = {}
zb_base = ZERO_BUFFERS[0]
for nd in ENTER_DAYS_LIST:
    results[f"Baseline: in {nd}↓ | hold to window end"] = run(nd, zb=zb_base)
    for tp in BASE_TPS:
        results[f"Baseline: in {nd}↓ | TP {pct(tp)}"] = run(nd, tp=tp, zb=zb_base)
for nd in ENTER_DAYS_LIST:
    for nu in EXIT_DAYS_LIST:
        for zb in ZERO_BUFFERS:
            for sk in SKIPS:
                for lb in (LOOKBACKS if sk is not None else [None]):
                    results[name(nd, nu, lb, sk, zb)] = run(nd, nu, lb, sk, zb=zb)
nd0, nu0, lb0, sk0, zb0 = SHOW
show = name(nd0, nu0, lb0 if sk0 is not None else None, sk0, zb0)
if show not in results:                                  # SHOW always runs, even outside the grid
    if sk0 is not None and (sk0[0], lb0) not in REFS:
        REFS[(sk0[0], lb0)] = (lowest_low if sk0[0] == "lowest" else recent_low)(lb0).loc[idx].values
    results[show] = run(nd0, nu0, lb0, sk0, zb=zb0)

# ------------------------- TABLES -------------------------------
types = pd.Series([w["type"] for w in windows]).value_counts()
print(f"{len(windows)} risk-off windows ({BASELINE} SMA{BASE_SMA} ±{pct(BASE_BUFFER)}): {types.get('shallow', 0)} shallow, "
      f"{types.get('flash', 0)} flash, {types.get('bear', 0)} bear | trade {TRADE} | PMO({PMO_LEN1},{PMO_LEN2}) | "
      f"swing lows K={PIVOT_K}{' below 0' if PIVOT_BELOW_ZERO else ''} | stop loss {pct(STOP_LOSS) if STOP_LOSS else 'off'}")
summary = pd.DataFrame({n: summarize(r) for n, r in results.items()}).T
cnt = ["Trades", "First entries skipped", "First entries blocked (PMO<0)"]
display(summary.style.format({c: ("{:.0f}" if c in cnt else "{:.1f}" if c == "Avg days held" else "{:.1%}")
                              for c in summary.columns}, na_rep="–"))

skip_rows = [(sk, lb) for sk in SKIPS for lb in (LOOKBACKS if sk is not None else [None])]
for nd in ENTER_DAYS_LIST:
    for metric in ["Median window", "% windows > 0", "Compounded ex-best"]:
        grid = pd.DataFrame({f"out {nu}↑ | zero ±{zb:g}": [summary.loc[name(nd, nu, lb, sk, zb), metric] for sk, lb in skip_rows]
                             for nu in EXIT_DAYS_LIST for zb in ZERO_BUFFERS},
                            index=[skip_label(sk) + ("" if sk is None else f" | lb {lb}") for sk, lb in skip_rows])
        display(grid.style.format("{:.1%}").background_gradient(cmap="RdYlGn", axis=None)
                .set_caption(f"in {nd}↓ — {metric}"))

res = results[show]
compare = [f"Baseline: in {nd0}↓ | hold to window end"] + [f"Baseline: in {nd0}↓ | TP {pct(tp)}" for tp in BASE_TPS] + \
          [name(nd0, nu0, None, None, zb0), show]
compare = list(dict.fromkeys(n for n in compare if n in results))
W = pd.DataFrame({"Start": [idx[w["s"]].date() for w in windows],
                  "End": [f"{idx[w['e']].date()}{' (open)' if w['open'] else ''}" for w in windows],
                  "Days": [w["e"] - w["s"] for w in windows], "Type": [w["type"] for w in windows],
                  "Max depth": [w["depth"] for w in windows], BASELINE: [B[w["e"]] / B[w["s"]] - 1 for w in windows]})
for n in compare:
    W[n.replace("Baseline: ", "").replace("PMO turn | ", "")] = [x["ret"] for x in results[n]["wins"]]
print(f"Per window — SHOW config: {show}")
display(W.style.format({c: "{:.1%}" for c in W.columns[4:]}))
if len(res["trades"]):
    log = res["trades"].copy()
    log["entry"], log["exit"] = log["entry"].dt.date, log["exit"].dt.date
    print(f"Trade log — {show}")
    display(log.style.format({"ret": "{:.1%}"}))

# ------------------------- PLOT ---------------------------------
type_color = {"shallow": "gold", "flash": "orange", "bear": "red"}
exit_color = {"PMO turn up": "blue", "PMO upper": "magenta", "take-profit": "green", "stop loss": "darkred",
              "window end": "black", "still open": "gray"}
pv_dates, pv_vals = pmo.index[piv_i], piv_v
skip_line = REFS[(sk0[0], lb0)] * sk0[1] if sk0 is not None and (sk0[0], lb0) in REFS else None
curve = res["curve"].values


def draw(a, z, title, rebase=False, figsize=(15, 15)):
    """Draw the 4 panels for bars a..z (inclusive)."""
    sl = slice(a, z + 1)
    x = idx[sl]
    fig, (ax1, ax2, axp, ax3) = plt.subplots(4, 1, figsize=figsize, sharex=True,
                                             gridspec_kw={"height_ratios": [3, 3, 2, 1.3]})
    fig.suptitle(title, fontsize=12, y=0.995)
    ax1.plot(x, B[sl], color="black", lw=1, label=BASELINE)
    ax1.plot(x, BS[sl], color="tab:blue", lw=1, label=f"SMA {BASE_SMA}")
    ax1.plot(x, BS[sl] * (1 - BASE_BUFFER), color="tab:blue", lw=0.7, ls="--", label=f"risk-off line -{pct(BASE_BUFFER)}")
    ax1.plot(x, BS[sl] * (1 + BASE_BUFFER), color="tab:blue", lw=0.7, ls=":", label=f"risk-on line +{pct(BASE_BUFFER)}")
    ax1.set_yscale("log"); ax1.set_title(f"{BASELINE} — windows shaded by type (yellow shallow / orange flash / red bear)")
    ax2.plot(x, T[sl], color="purple", lw=1, label=f"Trade {TRADE}")
    ax2.set_yscale("log")
    ax2.set_title(f"{show}  (▲ first, △ turn entry, ✕ red skipped / grey blocked PMO<0, ▼ exit by reason)")
    axp.plot(x, P[sl], color="black", lw=1, label="PMO")
    axp.axhline(0, color="gray", lw=0.8)
    if zb0: axp.axhspan(-zb0, zb0, color="gray", alpha=0.12, label=f"zero band ±{zb0:g}")
    if PMO_UPPER_EXIT is not None:
        axp.axhline(PMO_UPPER_EXIT, color="magenta", lw=0.8, ls="--", label=f"upper exit {PMO_UPPER_EXIT:g}")
    if skip_line is not None:
        axp.plot(x, skip_line[sl], color="crimson", lw=1, ls="--",
                 label=f"skip line: {sk0[0]} swing low × {sk0[1]:g} ({lb0} bars)")
    m = (pv_dates >= x[0]) & (pv_dates <= x[-1])
    axp.scatter(pv_dates[m], pv_vals[m], s=12, color="crimson", zorder=3, label="swing lows")
    axp.set_title("PMO with swing lows, skip line, entries (▲) and exits (▼)")
    for w in windows:
        if w["e"] < a or w["s"] > z: continue
        for ax in (ax1, ax2, axp, ax3):
            ax.axvspan(idx[max(w["s"], a)], idx[min(w["e"], z)], color=type_color[w["type"]], alpha=0.12, lw=0)
    entry_t = None
    for kind, t, pr, why in res["marks"]:
        if kind == "entry": entry_t = t
        if not a <= t <= z: continue
        if kind == "entry":
            style = dict(color="purple") if why == "first" else dict(color="white", edgecolor="purple")
            for ax, y in ((ax2, pr), (ax1, B[t]), (axp, P[t])):
                ax.scatter(idx[t], y, marker="^", s=40, zorder=4, **style)
        elif kind in ("skip", "blocked"):
            for ax, y in ((ax2, pr), (axp, P[t])):
                ax.scatter(idx[t], y, marker="x", s=46, zorder=4, color="crimson" if kind == "skip" else "gray")
        else:
            for ax, y in ((ax2, pr), (ax1, B[t]), (axp, P[t])):
                ax.scatter(idx[t], y, marker="v", s=40, zorder=4, color=exit_color[why])
            if entry_t is not None:
                ax2.axvspan(idx[max(entry_t, a)], idx[t], color="gray", alpha=0.18, lw=0)
    for why, c in exit_color.items():
        ax2.scatter([], [], marker="v", color=c, label=f"exit: {why}")
    eqv = curve[sl] / curve[a] if rebase else curve[sl]
    ax3.plot(x, eqv, color="tab:green", lw=1.2,
             label="Sleeve equity, rebased to 1 at chart start" if rebase else "Sleeve equity (cash when flat)")
    if rebase: ax3.axhline(1, color="gray", lw=0.8)
    else: ax3.set_yscale("log")
    for ax in (ax1, ax2, axp, ax3): ax.grid(alpha=0.3); ax.legend(loc="upper left", fontsize=8, ncol=2)
    plt.tight_layout(); plt.show()


if PLOT_FULL:
    draw(0, N - 1, f"Full history — {show}")

if PLOT_WINDOWS:
    pad_a, pad_z = WINDOW_PAD
    for wn, w in enumerate(windows):
        if PLOT_WINDOW_TYPES and w["type"] not in PLOT_WINDOW_TYPES: continue
        tr_w = res["trades"][res["trades"]["window"] == wn + 1] if len(res["trades"]) else res["trades"]
        title = (f"Window {wn + 1}/{len(windows)}: {idx[w['s']].date()} → {idx[w['e']].date()}"
                 f"{' (open)' if w['open'] else ''} | {w['type']} | {w['e'] - w['s']} days | "
                 f"max depth {w['depth']:.1%} | {BASELINE} {B[w['e']] / B[w['s']] - 1:+.1%} | "
                 f"window return {res['wins'][wn]['ret']:+.1%} | {len(tr_w)} trades")
        draw(max(0, w["s"] - pad_a), min(N - 1, w["e"] + pad_z), title, rebase=True, figsize=(15, 12))
```

```python
# ==============================================================
# Risk-OFF PMO Swing-Low Exit Tester  (standalone Jupyter cell)
# --------------------------------------------------------------
# Window      : BASELINE closes below SMA*(1-BASE_BUFFER) ... until it closes above SMA*(1+BASE_BUFFER)
# First entry : first close in the window with BASELINE below SMA*(1-ENTRY_BUFFER)
# Re-entries  : REENTRY_MODE 'inv_sma' (default): TRADE closes above its SMA*(1+INV_BUFFER) after closing below
#               SMA*(1-INV_BUFFER) since the last entry (hysteresis); 'pmo': PMO crosses DOWN through -ZERO_BUFFER
# PMO         : DecisionPoint PMO on BASELINE = EMA(2/LEN2) of 10*EMA(2/LEN1) of 1-bar ROC%
# Swing low   : a PMO value that is the lowest of the PIVOT_K bars either side (optionally only below zero),
#               usable only once confirmed, i.e. PIVOT_K bars later (no look-ahead)
# Reference   : the lowest confirmed swing low within the last LOOKBACK bars (frozen at entry, or rolling)
# PMO exit    : sell at the close when PMO <= reference * (1 + PMO_BUFFER)   ("a new lower low")
# Zone exit   : "higher low" — zone = zone-reference * ZONE_RATIO (reference = lowest swing low in the lookback,
#               or the most recent swing low). Once PMO has been in the zone during the trade without breaching,
#               sell at the close when PMO has risen ZONE_TURN_DAYS closes in a row. Zone level frozen at entry.
# Skip rule   : don't enter if PMO is already at/below the breach threshold, or (SKIP_ZONE_RISING) already in the
#               zone and rising — the trough has already happened. A skipped first entry falls through to re-entries.
# Zero exit   : PMO closes above ZERO_BUFFER (e.g. -0.1) -> sell at the close; live once PMO has been at/below
#               that level since entry
# Fallbacks   : zero exit, window end, optional stop loss
# Baselines   : same entries / zero exit / re-entries, no skip rule — PMO zero only, fixed TPs, RSI cross
# Fills       : entries / close-based exits at the close; stops & TPs on daily open/high/low
# ==============================================================
import numpy as np
import pandas as pd
import yfinance as yf
import matplotlib.pyplot as plt
from IPython.display import display

# ------------------------- INPUTS -------------------------------
BASELINE       = "SPY"
BASE_SMA       = 200
BASE_BUFFER    = 0.04
ENTRY_BUFFER   = 0.00                 # = BASE_BUFFER -> enter the day the window opens
TRADE          = {"SPY": -3.0}        # traded inverse (synthetic SPXU); real ETF: {"SPXU": 1}
TRADE_COST     = 0.0
REENTRY_MODE   = "inv_sma"            # "inv_sma" (TRADE crosses back above its SMA) or "pmo" (PMO crosses below -zero buffer)
INV_SMA        = 200
INV_BUFFER     = 0.03                 # hysteresis band on TRADE's SMA for re-entries
REENTRY_NEEDS_DEPTH = False           # re-entries also need BASELINE below SMA*(1-ENTRY_BUFFER)
STOP_LOSS      = None                 # e.g. 0.15; None = off (applies to every config)

PMO_LEN1, PMO_LEN2 = 35, 20
PIVOT_K        = 5                    # swing low = lowest of K bars either side, confirmed K bars later
PIVOT_BELOW_ZERO = True               # only troughs below zero count as swing lows
LOOKBACKS      = [252, 252*1.5]           # bars to look back for swing lows (~6m / 1y); 504 = ~2y
PMO_BUFFERS    = [0.0, 0.05]          # breach threshold = lowest swing low * (1 + buffer)  (more negative)
REF_FROZEN     = True                 # True: threshold fixed at entry; False: updates daily
SKIP_OPTIONS   = [True]               # [True, False] to also test without the skip rule
ZERO_BUFFERS   = [-0.1, -0.2]               # zero exit fires when PMO closes above this (PMO units); list to test several

ZONE_RATIOS    = [None, 0.5, 0.65]   # zone starts at reference * ratio; None = no zone exit
ZONE_TURN_DAYS = [2]                   # consecutive rising PMO closes (inside/after the zone) to sell
ZONE_REFS      = ["lowest", "recent"]     # zone reference: lowest swing low in lookback, or most recent swing low
SKIP_ZONE_RISING = True                   # skip entries made while PMO is already in the zone and rising

BASE_TPS       = [0.04]         # baseline: fixed TP only
RSI_PERIOD, RSI_EXIT = 14, 30         # baseline: sell at close when SPY RSI crosses below RSI_EXIT

SHALLOW_DEPTH, FLASH_DAYS = 0.08, 90  # hindsight window tags (analysis only)
CASH_RATE      = 0.03
START, END     = "1994-01-01", "2026-10-22"
SHOW           = (252*1.5, 0.05, True, -0.2, 0.65, 2, "nearest")
                 # (LOOKBACK, PMO_BUFFER, skip, ZERO_BUFFER, ZONE_RATIO, ZONE_TURN_DAYS, ZONE_REF) for tables + chart
TRADING_DAYS   = 252
# ----------------------------------------------------------------

pct = lambda x: f"{x * 100:g}%"
tickers = sorted({BASELINE, *TRADE})
warmup = pd.Timestamp(START) - pd.Timedelta(days=int(max(BASE_SMA, INV_SMA, max(LOOKBACKS)) * 1.6) + 200)
raw = yf.download(tickers, start=warmup, end=END, auto_adjust=True, progress=False)
if not isinstance(raw.columns, pd.MultiIndex):
    raw.columns = pd.MultiIndex.from_product([raw.columns, tickers])
raw = raw[["Open", "High", "Low", "Close"]].dropna(how="any")
px = raw["Close"]
rets = px.pct_change()

def synth(weights, cost=0.0):
    w = pd.Series(weights, dtype=float)
    r = rets[w.index].mul(w, axis=1).clip(lower=-w.abs(), axis=1).sum(axis=1)
    r = (r - cost / TRADING_DAYS).clip(lower=-1.0); r.iloc[0] = 0.0
    return (1 + r).cumprod()

def synth_ohl(weights, close):
    w = pd.Series(weights, dtype=float)
    prev = px[w.index].shift(1)
    move = lambda f: raw[f][w.index] / prev - 1
    o = (move("Open") * w).sum(axis=1)
    hi = sum(wt * move("High" if wt > 0 else "Low")[t] for t, wt in w.items())
    lo = sum(wt * move("Low" if wt > 0 else "High")[t] for t, wt in w.items())
    base = close.shift(1)
    return base * (1 + o).clip(lower=0), base * (1 + hi), base * (1 + lo).clip(lower=0)

trd = synth(TRADE, TRADE_COST)
trd_o, trd_h, trd_l = synth_ohl(TRADE, trd)
trd_sma = trd.rolling(INV_SMA).mean()

b = px[BASELINE]
b_sma = b.rolling(BASE_SMA).mean()
g, cur = np.full(len(b), np.nan), None
for k_, (p, s) in enumerate(zip(b.values, b_sma.values)):
    if np.isnan(s): continue
    if cur is None: cur = p > s
    elif cur and p < s * (1 - BASE_BUFFER): cur = False
    elif not cur and p > s * (1 + BASE_BUFFER): cur = True
    g[k_] = cur
gate = pd.Series(g, index=b.index)

# PMO + RSI on the baseline
roc = b.pct_change() * 100
pmo = (10 * roc.ewm(alpha=2 / PMO_LEN1, adjust=False).mean()).ewm(alpha=2 / PMO_LEN2, adjust=False).mean()
d_ = b.diff()
rsi = 100 - 100 / (1 + d_.clip(lower=0).ewm(alpha=1 / RSI_PERIOD, adjust=False).mean()
                   / (-d_.clip(upper=0)).ewm(alpha=1 / RSI_PERIOD, adjust=False).mean())

# swing lows (on the full history) and the reference low available on each day
Pf = pmo.values
piv_i = np.array([i for i in range(PIVOT_K, len(Pf) - PIVOT_K)
                  if not np.isnan(Pf[i]) and Pf[i] == np.nanmin(Pf[i - PIVOT_K:i + PIVOT_K + 1])
                  and (Pf[i] < 0 or not PIVOT_BELOW_ZERO)], dtype=int)
piv_v, piv_c = Pf[piv_i], piv_i + PIVOT_K            # value, confirmation bar

def reference_low(lookback):
    """Lowest confirmed swing low with pivot bar in [t-lookback, t] and confirmed by t; NaN if none."""
    ref = np.full(len(Pf), np.nan)
    if len(piv_i) == 0: return pd.Series(ref, index=pmo.index)
    tt = np.arange(len(Pf))[:, None]
    ok = (piv_c[None, :] <= tt) & (piv_i[None, :] >= tt - lookback)
    vals = np.where(ok, piv_v[None, :], np.inf).min(axis=1)
    ref[np.isfinite(vals)] = vals[np.isfinite(vals)]
    return pd.Series(ref, index=pmo.index)

def recent_low(lookback):
    """Most recent confirmed swing low (pivot bar within the lookback); NaN if none."""
    ref = np.full(len(Pf), np.nan)
    if len(piv_i) == 0: return pd.Series(ref, index=pmo.index)
    tt = np.arange(len(Pf))
    j = np.searchsorted(piv_c, tt, side="right") - 1
    ok = j >= 0
    ok[ok] &= piv_i[j[ok]] >= tt[ok] - lookback
    ref[ok] = piv_v[j[ok]]
    return pd.Series(ref, index=pmo.index)

refs = {lb: reference_low(lb) for lb in LOOKBACKS}
recents = {lb: recent_low(lb) for lb in LOOKBACKS}

idx = px.index[(px.index >= pd.Timestamp(START)) & gate.notna() & trd_sma.notna()]
B, BS = b.loc[idx].values, b_sma.loc[idx].values
T, TO, TH, TL = (x.loc[idx].values for x in (trd, trd_o, trd_h, trd_l))
P, R = pmo.loc[idx].values, rsi.loc[idx].values
SS = trd_sma.loc[idx].values
S_HI, S_LO = SS * (1 + INV_BUFFER), SS * (1 - INV_BUFFER)
REF = {lb: refs[lb].loc[idx].values for lb in LOOKBACKS}
REC = {lb: recents[lb].loc[idx].values for lb in LOOKBACKS}
ON = gate.loc[idx].astype(bool).values
N = len(idx)

windows, i = [], 1
while i < N:
    if ON[i-1] and not ON[i]:
        j = i
        while j < N and not ON[j]: j += 1
        e = min(j, N - 1)
        depth = (B[i:e+1] / BS[i:e+1] - 1).min()
        kind = "shallow" if depth > -SHALLOW_DEPTH else ("flash" if e - i <= FLASH_DAYS else "bear")
        windows.append(dict(s=i, e=e, open=j >= N, depth=depth, type=kind))
        i = j
    i += 1
if not windows:
    raise ValueError("No risk-off windows in this date range.")
win_of = np.full(N, -1)
for wn, w in enumerate(windows):
    win_of[w["s"]:w["e"] + 1] = wn


def threshold(lb, buf, t):
    r = REF[lb][t]
    return r * (1 + buf) if not np.isnan(r) and r < 0 else np.nan


def zone_level(lb, zr, zref, t):
    if zr is None or lb is None: return np.nan
    r = (REF if zref == "lowest" else REC)[lb][t]
    return r * zr if not np.isnan(r) and r < 0 else np.nan


def run(mode, zb, lb=None, buf=None, skip=False, tp=None, zr=None, zn=1, zref="lowest"):
    """mode: 'pmo' (lower-low + optional zone exit) | 'zero' (zero exit only) | 'tp' | 'rsi'; zb = zero-exit level"""
    cash_d = (1 + CASH_RATE) ** (1 / TRADING_DAYS) - 1
    eq, curve, trades, marks = 1.0, np.ones(N), [], []
    in_pos, st, entered, reset, skips = False, {}, False, False, 0

    for t in range(1, N):
        wn = win_of[t]
        if wn >= 0 and t == windows[wn]["s"]:
            entered, reset = False, False
        exited_today = False

        if in_pos:
            ep, hit, stops = st["ep"], None, []
            if STOP_LOSS:
                L = ep * (1 - STOP_LOSS)
                if TO[t] <= L: stops.append((TO[t], "stop loss"))
                elif TL[t] <= L: stops.append((L, "stop loss"))
            if mode == "tp":
                L = ep * (1 + tp)
                if TO[t] >= L: hit = (TO[t], "take-profit")
                elif TH[t] >= L and not stops: hit = (L, "take-profit")
            if hit is None and stops: hit = max(stops)
            st["up"] = st["up"] + 1 if P[t] > P[t-1] else 0
            if not np.isnan(st["zl"]) and P[t] <= st["zl"]: st["in_zone"] = True
            if hit is None and mode == "pmo":
                th = st["th"] if REF_FROZEN else threshold(lb, buf, t)
                if not np.isnan(th) and P[t] <= th: hit = (T[t], "PMO low")
            if hit is None and mode == "pmo" and st["in_zone"] and st["up"] >= zn:
                hit = (T[t], "PMO zone turn")
            if hit is None and mode == "rsi" and R[t-1] >= RSI_EXIT and R[t] < RSI_EXIT:
                hit = (T[t], "RSI")
            if hit is None and st["zero_armed"] and P[t] > zb:
                hit = (T[t], "PMO zero")
            if hit is None and ON[t]: hit = (T[t], "window end")
            if hit:
                fill, why = hit
                eq *= fill / T[t-1]
                trades.append(dict(window=st["wn"] + 1, kind=st["kind"], entry=idx[st["t0"]], exit=idx[t],
                                   days=t - st["t0"], ret=fill / ep - 1, reason=why))
                marks.append(("exit", t, fill, why))
                in_pos, exited_today = False, True
            else:
                eq *= T[t] / T[t-1]
                st["zero_armed"] = st["zero_armed"] or P[t] <= zb
        else:
            eq *= 1 + cash_d

        if not in_pos and wn >= 0 and not ON[t]:
            deep = B[t] <= BS[t] * (1 - ENTRY_BUFFER)
            kind = None
            if not entered and deep:
                kind = "first"
            elif entered and not exited_today and (deep or not REENTRY_NEEDS_DEPTH) and (
                    (REENTRY_MODE == "inv_sma" and reset and T[t] > S_HI[t]) or
                    (REENTRY_MODE == "pmo" and P[t-1] > -zb and P[t] <= -zb)):
                kind = "re-entry"
            if kind:
                entered, reset = True, False
                th = threshold(lb, buf, t) if mode == "pmo" else np.nan
                zl = zone_level(lb, zr, zref, t) if mode == "pmo" else np.nan
                late = (not np.isnan(th) and P[t] <= th) or \
                       (SKIP_ZONE_RISING and not np.isnan(zl) and P[t] <= zl and P[t] > P[t-1])
                if skip and late:
                    skips += 1
                    marks.append(("skip", t, T[t], kind))
                else:
                    in_pos = True
                    st = dict(ep=T[t], t0=t, wn=wn, kind=kind, zero_armed=P[t] <= zb, th=th, zl=zl,
                              in_zone=not np.isnan(zl) and P[t] <= zl, up=0)
                    marks.append(("entry", t, T[t], kind))
            elif entered and T[t] <= S_LO[t]:
                reset = True
        curve[t] = eq

    if in_pos:
        trades.append(dict(window=st["wn"] + 1, kind=st["kind"], entry=idx[st["t0"]], exit=idx[-1],
                           days=N - 1 - st["t0"], ret=T[-1] / st["ep"] - 1, reason="still open"))
    curve = pd.Series(curve, index=idx)
    wins = [dict(ret=curve.iloc[w["e"]] / curve.iloc[w["s"] - 1] - 1,
                 traded=any(m[0] == "entry" and win_of[m[1]] == wn for m in marks)) for wn, w in enumerate(windows)]
    return dict(curve=curve, trades=pd.DataFrame(trades), marks=marks, wins=wins, skips=skips)


REASONS = ["PMO low", "PMO zone turn", "RSI", "take-profit", "PMO zero", "stop loss", "window end", "still open"]

def summarize(res):
    w = pd.DataFrame(res["wins"]).assign(type=[x["type"] for x in windows])
    tr, c, r = res["trades"], res["curve"], np.array([x["ret"] for x in res["wins"]])
    years = (c.index[-1] - c.index[0]).days / 365.25
    out = {"Windows traded": w["traded"].mean(), "Mean window": r.mean(), "Median window": np.median(r),
           "% windows > 0": (r > 0).mean(), "% windows ≥ 10%": (r >= 0.10).mean(),
           "Worst window": r.min(), "Best window": r.max(),
           "Compounded": np.prod(1 + r) - 1, "Compounded ex-best": np.prod(1 + np.sort(r)[:-1]) - 1}
    for k in ["shallow", "flash", "bear"]:
        sub = w[w.type == k]
        out[f"Mean {k} ({len(sub)})"] = sub["ret"].mean() if len(sub) else np.nan
    has = len(tr) > 0
    out.update({"Trades": len(tr), "Entries skipped": res["skips"],
                "Trade win rate": (tr["ret"] > 0).mean() if has else np.nan,
                "Avg trade": tr["ret"].mean() if has else np.nan,
                "Avg days held": tr["days"].mean() if has else np.nan,
                "Sleeve CAGR": c.iloc[-1] ** (1 / years) - 1, "Sleeve max DD": (c / c.cummax() - 1).min()})
    why = tr["reason"].value_counts(normalize=True) if has else pd.Series(dtype=float)
    out.update({f"Exit: {k}": why.get(k, 0.0) for k in REASONS})
    return out


zone_label = lambda zr, zn, zref: "zone off" if zr is None else f"zone {zr:g}×{zref} {zn}d"
zone_cfgs = [(None, 1, "lowest")] + [(zr, zn, zref) for zr in ZONE_RATIOS if zr is not None
                                     for zn in ZONE_TURN_DAYS for zref in ZONE_REFS]
name_pmo = lambda lb, buf, sk, zb, zr, zn, zref: (f"PMO | lb {lb} | buf {pct(buf)}{'' if sk else ' | no skip'} | "
                                                  f"{zone_label(zr, zn, zref)} | zero>{zb:g}")
base_names = lambda zb: ([f"Baseline: PMO zero only | zero>{zb:g}"] +
                         [f"Baseline: TP {pct(tp)} | zero>{zb:g}" for tp in BASE_TPS] +
                         [f"Baseline: RSI<{RSI_EXIT:g} | zero>{zb:g}"])
results = {}
for zb in ZERO_BUFFERS:
    nm = base_names(zb)
    results[nm[0]] = run("zero", zb)
    for tp, n_ in zip(BASE_TPS, nm[1:-1]):
        results[n_] = run("tp", zb, tp=tp)
    results[nm[-1]] = run("rsi", zb)
for zb in ZERO_BUFFERS:
    for sk in SKIP_OPTIONS:
        for lb in LOOKBACKS:
            for buf in PMO_BUFFERS:
                for zr, zn, zref in zone_cfgs:
                    results[name_pmo(lb, buf, sk, zb, zr, zn, zref)] = run("pmo", zb, lb, buf, sk, zr=zr, zn=zn, zref=zref)

# ------------------------- TABLES -------------------------------
types = pd.Series([w["type"] for w in windows]).value_counts()
print(f"{len(windows)} risk-off windows ({BASELINE} SMA{BASE_SMA} ±{pct(BASE_BUFFER)}, entry -{pct(ENTRY_BUFFER)}): "
      f"{types.get('shallow', 0)} shallow, {types.get('flash', 0)} flash, {types.get('bear', 0)} bear | "
      f"trade {TRADE} | PMO({PMO_LEN1},{PMO_LEN2}) swing lows K={PIVOT_K}{' below 0' if PIVOT_BELOW_ZERO else ''} | "
      f"threshold {'frozen at entry' if REF_FROZEN else 'rolling'} | stop loss {pct(STOP_LOSS) if STOP_LOSS else 'off'} | "
      f"zero exit: PMO above {', '.join(format(z, 'g') for z in ZERO_BUFFERS)} | re-entry: "
      f"{f'TRADE SMA{INV_SMA} ±{pct(INV_BUFFER)}' if REENTRY_MODE == 'inv_sma' else 'PMO below -zero buffer'}")
summary = pd.DataFrame({n: summarize(r) for n, r in results.items()}).T
cnt_cols = ["Trades", "Entries skipped"]
display(summary.style.format({c: ("{:.0f}" if c in cnt_cols else "{:.1f}" if c == "Avg days held" else "{:.1%}")
                              for c in summary.columns}, na_rep="–"))

pmo_rows = [n for n in summary.index if n.startswith("PMO |")]
top_cols = ["Median window", "Mean window", "% windows > 0", "% windows ≥ 10%", "Worst window", "Compounded ex-best",
            "Trades", "Trade win rate", "Exit: PMO low", "Exit: PMO zone turn", "Exit: PMO zero", "Exit: window end"]
top = summary.loc[pmo_rows].sort_values(["Median window", "% windows > 0"], ascending=False).head(15)[top_cols]
display(top.style.format({c: ("{:.0f}" if c == "Trades" else "{:.1%}") for c in top_cols})
        .set_caption("Top 15 PMO configs by median window"))

for zb in ZERO_BUFFERS:
    for sk in SKIP_OPTIONS:
        for metric in ["Median window", "% windows > 0", "Compounded ex-best"]:
            grid = pd.DataFrame({f"lb {lb} | buf {pct(buf)}":
                                 [summary.loc[name_pmo(lb, buf, sk, zb, *z), metric] for z in zone_cfgs]
                                 for lb in LOOKBACKS for buf in PMO_BUFFERS},
                                index=[zone_label(*z) for z in zone_cfgs])
            display(grid.style.format("{:.1%}").background_gradient(cmap="RdYlGn", axis=None)
                    .set_caption(f"{metric} — zero>{zb:g}{'' if sk else ', no skip'}"))

lb0, buf0, sk0, zb0, zr0, zn0, zref0 = SHOW
show = name_pmo(*SHOW) if name_pmo(*SHOW) in results else pmo_rows[0]
if show != name_pmo(*SHOW):
    print(f"SHOW config not in the grid — showing {show}")
    lb0, buf0, sk0, zb0 = LOOKBACKS[0], PMO_BUFFERS[0], SKIP_OPTIONS[0], ZERO_BUFFERS[0]
    zr0, zn0, zref0 = zone_cfgs[0]
res = results[show]
compare = base_names(zb0) + [show]
W = pd.DataFrame({"Start": [idx[w["s"]].date() for w in windows],
                  "End": [f"{idx[w['e']].date()}{' (open)' if w['open'] else ''}" for w in windows],
                  "Days": [w["e"] - w["s"] for w in windows], "Type": [w["type"] for w in windows],
                  "Max depth": [w["depth"] for w in windows],
                  BASELINE: [B[w["e"]] / B[w["s"]] - 1 for w in windows]})
for n in compare:
    W[n.replace("Baseline: ", "").split(" | zero>")[0] if n != show else "Shown PMO config"] = \
        [x["ret"] for x in results[n]["wins"]]
print(f"Per window — shown config: {show}")
display(W.style.format({c: "{:.1%}" for c in W.columns[4:]}))
if len(res["trades"]):
    log = res["trades"].copy()
    log["entry"], log["exit"] = log["entry"].dt.date, log["exit"].dt.date
    print(f"Trade log — {show}")
    display(log.style.format({"ret": "{:.1%}"}))

# ------------------------- PLOT ---------------------------------
type_color = {"shallow": "gold", "flash": "orange", "bear": "red"}
exit_color = {"PMO low": "blue", "PMO zone turn": "cyan", "RSI": "teal", "take-profit": "green", "PMO zero": "orange",
              "stop loss": "darkred", "window end": "black", "still open": "gray"}
fig, (ax1, ax2, axp, ax3) = plt.subplots(4, 1, figsize=(15, 15), sharex=True,
                                         gridspec_kw={"height_ratios": [3, 3, 2, 1.3]})
ax1.plot(idx, B, color="black", lw=1, label=BASELINE)
ax1.plot(idx, BS, color="tab:blue", lw=1, label=f"SMA {BASE_SMA}")
ax1.plot(idx, BS * (1 - BASE_BUFFER), color="tab:blue", lw=0.7, ls="--", label=f"risk-off line -{pct(BASE_BUFFER)}")
ax1.set_yscale("log"); ax1.set_title(f"{BASELINE} — windows shaded by type (yellow shallow / orange flash / red bear)")
ax2.plot(idx, T, color="purple", lw=1, label=f"Trade {TRADE}")
ax2.plot(idx, SS, color="orange", lw=0.9, label=f"Trade SMA {INV_SMA} (re-entries)")
ax2.set_yscale("log")
ax2.set_title(f"{show}  (▲ first, △ re-entry, ✕ skipped, ▼ exit by reason; grey = in trade)")
axp.plot(idx, P, color="black", lw=1, label="PMO")
axp.plot(idx, REF[lb0] * (1 + buf0), color="crimson", lw=1, ls="--",
         label=f"threshold: lowest swing low ({lb0} bars) × {1 + buf0:g}")
pv_in = [k for k in range(len(piv_i)) if piv_i[k] < len(pmo) and pmo.index[piv_i[k]] >= idx[0]]
axp.scatter(pmo.index[piv_i[pv_in]], piv_v[pv_in], s=12, color="crimson", zorder=3, label="swing lows")
axp.axhline(0, color="gray", lw=0.8)
axp.axhline(zb0, color="orange", lw=0.8, ls=":", label=f"zero exit {zb0:g}")
if zr0 is not None and lb0 in LOOKBACKS:
    axp.plot(idx, (REF if zref0 == "lowest" else REC)[lb0] * zr0, color="cyan", lw=1, ls="--",
             label=f"zone: {zref0} swing low × {zr0:g}")
if REENTRY_MODE == "pmo":
    axp.axhline(-zb0, color="purple", lw=0.8, ls=":", label=f"re-entry -{zb0:g}")
axp.set_title("PMO on the baseline with swing lows and the exit threshold")
for w in windows:
    for ax in (ax1, ax2, axp, ax3):
        ax.axvspan(idx[w["s"]], idx[w["e"]], color=type_color[w["type"]], alpha=0.12, lw=0)
entry_t = None
for kind, t, p, why in res["marks"]:
    if kind == "entry":
        entry_t = t
        style = dict(color="purple") if why == "first" else dict(color="white", edgecolor="purple")
        ax2.scatter(idx[t], p, marker="^", s=40, zorder=4, **style)
        ax1.scatter(idx[t], B[t], marker="^", s=30, zorder=4, **style)
        axp.scatter(idx[t], P[t], marker="^", s=30, zorder=4, **style)
    elif kind == "skip":
        ax2.scatter(idx[t], p, marker="x", s=40, zorder=4, color="crimson")
        axp.scatter(idx[t], P[t], marker="x", s=40, zorder=4, color="crimson")
    else:
        ax2.scatter(idx[t], p, marker="v", s=40, zorder=4, color=exit_color[why])
        ax1.scatter(idx[t], B[t], marker="v", s=30, zorder=4, color=exit_color[why])
        if why in ("PMO low", "PMO zone turn", "PMO zero"): axp.scatter(idx[t], P[t], marker="v", s=40, zorder=4, color=exit_color[why])
        ax2.axvspan(idx[entry_t], idx[t], color="gray", alpha=0.18, lw=0)
for why, c in exit_color.items():
    ax2.scatter([], [], marker="v", color=c, label=f"exit: {why}")
ax3.plot(res["curve"].index, res["curve"].values, color="tab:green", lw=1.2, label="Sleeve equity (cash when flat)")
ax3.set_yscale("log")
for ax in (ax1, ax2, axp, ax3): ax.grid(alpha=0.3); ax.legend(loc="upper left", fontsize=8, ncol=2)
plt.tight_layout(); plt.show()
```

```python
# ==============================================================
# Risk-OFF PMO Turn Tester  (standalone Jupyter cell)
# --------------------------------------------------------------
# Window   : BASELINE closes below SMA*(1-BASE_BUFFER) ... until it closes above SMA*(1+BASE_BUFFER)
# PMO      : DecisionPoint PMO on BASELINE = EMA(2/LEN2) of 10*EMA(2/LEN1) of 1-bar ROC%
# First entry: the close the window opens (SPY breaks SMA - buffer), only if not skipped AND PMO > -ZERO_BUFFER.
#            Otherwise no first entry; the window waits for a turn entry.
# Turn entry : inside a window, while flat: PMO has fallen ENTER_DAYS closes in a row ("double negative") AND
#            PMO > -ZERO_BUFFER (buy only while momentum is still around/above zero). Never skipped.
# Exits    : turn exit — PMO has risen EXIT_DAYS closes in a row AND PMO < +ZERO_BUFFER (sell only below zero)
#            upper barrier — PMO closes at/above PMO_UPPER_EXIT
#            window end — sold at the close the day BASELINE turns risk-on; optional hard stop loss (intraday)
# Skip     : applies to the FIRST entry of a window only. Swing lows are used only for this. Swing low = PMO value lowest of PIVOT_K bars either
#            side (below zero), usable once confirmed PIVOT_K bars later. Reference = lowest swing low in the
#            lookback, or the most recent one. Skip the entry if PMO <= reference * ratio:
#              ratio 1.1 = already 10% past the previous low, 1.0 = at it, 0.8 / 0.65 = already near it.
# Fills    : entries and PMO / window-end exits at the close; stop loss on daily open/low
# ==============================================================
import numpy as np
import pandas as pd
import yfinance as yf
import matplotlib.pyplot as plt
from IPython.display import display

# ------------------------- INPUTS -------------------------------
BASELINE       = "SPY"
BASE_SMA       = 200
BASE_BUFFER    = 0.03
TRADE          = {"SPY": -3.0}        # traded inverse (synthetic SPXU); real ETF: {"SPXU": 1}
TRADE_COST     = 0.0                  # annual drag on TRADE (e.g. 0.03)
STOP_LOSS      = None                 # e.g. 0.15; None = off

PMO_LEN1, PMO_LEN2 = 35, 20
ENTER_DAYS_LIST = [2]                 # falling PMO closes in a row to enter
EXIT_DAYS_LIST  = [2, 3]              # rising PMO closes in a row to exit
ZERO_BUFFERS    = [0.0, 0.5]          # entries need PMO > -b, turn exits need PMO < +b (0 = strictly above/below 0)
PMO_UPPER_EXIT  = 2.2                 # sell when PMO closes at/above this; None = off

PIVOT_K        = 5
PIVOT_BELOW_ZERO = True
LOOKBACKS      = [252, 378]           # ~1 year / ~1.5 years
SKIPS          = [None, ("lowest", 1.1), ("lowest", 1.0), ("recent", 0.8), ("recent", 0.65)]
                 # (reference, ratio): skip if PMO <= reference * ratio; None = never skip

BASE_TPS       = [0.04, 0.05]         # baselines on the same entries (first ZERO_BUFFER, no skip): fixed TP + window end

SHALLOW_DEPTH, FLASH_DAYS = 0.08, 90  # hindsight window tags (analysis only)
CASH_RATE      = 0.03
START, END     = "2010-01-01", "2012-10-23"
SHOW           = (2, 2, 252, ("lowest", 1.1), 0.0)
                 # (ENTER_DAYS, EXIT_DAYS, LOOKBACK, SKIP, ZERO_BUFFER) — always run, even if not in the grid
TRADING_DAYS   = 252
# ----------------------------------------------------------------

pct = lambda x: f"{x * 100:g}%"
tickers = sorted({BASELINE, *TRADE})
warmup = pd.Timestamp(START) - pd.Timedelta(days=int(max(BASE_SMA, max(LOOKBACKS)) * 1.6) + 200)
raw = yf.download(tickers, start=warmup, end=END, auto_adjust=True, progress=False)
if not isinstance(raw.columns, pd.MultiIndex):
    raw.columns = pd.MultiIndex.from_product([raw.columns, tickers])
raw = raw[["Open", "High", "Low", "Close"]].dropna(how="any")
px = raw["Close"]
rets = px.pct_change()

def synth(weights, cost=0.0):
    w = pd.Series(weights, dtype=float)
    r = rets[w.index].mul(w, axis=1).clip(lower=-w.abs(), axis=1).sum(axis=1)
    r = (r - cost / TRADING_DAYS).clip(lower=-1.0); r.iloc[0] = 0.0
    return (1 + r).cumprod()

def synth_ohl(weights, close):
    w = pd.Series(weights, dtype=float)
    prev = px[w.index].shift(1)
    move = lambda f: raw[f][w.index] / prev - 1
    o = (move("Open") * w).sum(axis=1)
    hi = sum(wt * move("High" if wt > 0 else "Low")[t] for t, wt in w.items())
    lo = sum(wt * move("Low" if wt > 0 else "High")[t] for t, wt in w.items())
    base = close.shift(1)
    return base * (1 + o).clip(lower=0), base * (1 + hi), base * (1 + lo).clip(lower=0)

trd = synth(TRADE, TRADE_COST)
trd_o, trd_h, trd_l = synth_ohl(TRADE, trd)

b = px[BASELINE]
b_sma = b.rolling(BASE_SMA).mean()
g, cur = np.full(len(b), np.nan), None
for k_, (p, s) in enumerate(zip(b.values, b_sma.values)):
    if np.isnan(s): continue
    if cur is None: cur = p > s
    elif cur and p < s * (1 - BASE_BUFFER): cur = False
    elif not cur and p > s * (1 + BASE_BUFFER): cur = True
    g[k_] = cur
gate = pd.Series(g, index=b.index)

roc = b.pct_change() * 100
pmo = (10 * roc.ewm(alpha=2 / PMO_LEN1, adjust=False).mean()).ewm(alpha=2 / PMO_LEN2, adjust=False).mean()

# swing lows and references (full history, no look-ahead)
Pf = pmo.values
piv_i = np.array([i for i in range(PIVOT_K, len(Pf) - PIVOT_K)
                  if not np.isnan(Pf[i]) and Pf[i] == np.nanmin(Pf[i - PIVOT_K:i + PIVOT_K + 1])
                  and (Pf[i] < 0 or not PIVOT_BELOW_ZERO)], dtype=int)
piv_v, piv_c = Pf[piv_i], piv_i + PIVOT_K

def lowest_low(lookback):
    ref = np.full(len(Pf), np.nan)
    if len(piv_i):
        tt = np.arange(len(Pf))[:, None]
        ok = (piv_c[None, :] <= tt) & (piv_i[None, :] >= tt - lookback)
        vals = np.where(ok, piv_v[None, :], np.inf).min(axis=1)
        ref[np.isfinite(vals)] = vals[np.isfinite(vals)]
    return pd.Series(ref, index=pmo.index)

def recent_low(lookback):
    ref = np.full(len(Pf), np.nan)
    if len(piv_i):
        tt = np.arange(len(Pf))
        j = np.searchsorted(piv_c, tt, side="right") - 1
        ok = j >= 0
        ok[ok] &= piv_i[j[ok]] >= tt[ok] - lookback
        ref[ok] = piv_v[j[ok]]
    return pd.Series(ref, index=pmo.index)

idx = px.index[(px.index >= pd.Timestamp(START)) & gate.notna()]
B, BS = b.loc[idx].values, b_sma.loc[idx].values
T, TO, TL = (x.loc[idx].values for x in (trd, trd_o, trd_l))
TH = trd_h.loc[idx].values
P = pmo.loc[idx].values
REFS = {("lowest", lb): lowest_low(lb).loc[idx].values for lb in LOOKBACKS}
REFS.update({("recent", lb): recent_low(lb).loc[idx].values for lb in LOOKBACKS})
ON = gate.loc[idx].astype(bool).values
N = len(idx)

# PMO direction streaks (close-to-close)
DN, UP = np.zeros(N, int), np.zeros(N, int)
for t in range(1, N):
    DN[t] = DN[t-1] + 1 if P[t] < P[t-1] else 0
    UP[t] = UP[t-1] + 1 if P[t] > P[t-1] else 0

windows, i = [], 1
while i < N:
    if ON[i-1] and not ON[i]:
        j = i
        while j < N and not ON[j]: j += 1
        e = min(j, N - 1)
        depth = (B[i:e+1] / BS[i:e+1] - 1).min()
        kind = "shallow" if depth > -SHALLOW_DEPTH else ("flash" if e - i <= FLASH_DAYS else "bear")
        windows.append(dict(s=i, e=e, open=j >= N, depth=depth, type=kind))
        i = j
    i += 1
if not windows:
    raise ValueError("No risk-off windows in this date range.")
win_of = np.full(N, -1)
for wn, w in enumerate(windows):
    win_of[w["s"]:w["e"] + 1] = wn


def skip_level(lb, skip, t):
    if skip is None: return np.nan
    r = REFS[(skip[0], lb)][t]
    return r * skip[1] if not np.isnan(r) and r < 0 else np.nan


def run(nd, nu=None, lb=None, skip=None, tp=None, zb=0.0):
    """nd = falling days to enter, nu = rising days to exit (None = baseline: no PMO exits), tp = fixed TP,
    zb = zero buffer."""
    cash_d = (1 + CASH_RATE) ** (1 / TRADING_DAYS) - 1
    eq, curve, trades, marks = 1.0, np.ones(N), [], []
    in_pos, st, skips, n_entries, blocked = False, {}, 0, 0, 0

    for t in range(1, N):
        wn = win_of[t]
        exited_today = False
        if in_pos:
            ep, hit = st["ep"], None
            if STOP_LOSS:
                L = ep * (1 - STOP_LOSS)
                if TO[t] <= L: hit = (TO[t], "stop loss")
                elif TL[t] <= L: hit = (L, "stop loss")
            if hit is None and tp is not None:
                L = ep * (1 + tp)
                if TO[t] >= L: hit = (TO[t], "take-profit")
                elif TH[t] >= L: hit = (L, "take-profit")
            if hit is None and nu and UP[t] >= nu and t - st["t0"] >= nu and P[t] < zb:
                hit = (T[t], "PMO turn up")
            if hit is None and nu and PMO_UPPER_EXIT is not None and P[t] >= PMO_UPPER_EXIT:
                hit = (T[t], "PMO upper")
            if hit is None and ON[t]: hit = (T[t], "window end")
            if hit:
                fill, why = hit
                eq *= fill / T[t-1]
                trades.append(dict(window=st["wn"] + 1, kind=st["kind"], entry=idx[st["t0"]], exit=idx[t],
                                   days=t - st["t0"], ret=fill / ep - 1, reason=why))
                marks.append(("exit", t, fill, why))
                in_pos, exited_today = False, True
            else:
                eq *= T[t] / T[t-1]
        else:
            eq *= 1 + cash_d

        if not in_pos and not exited_today and wn >= 0 and not ON[t]:
            kind = None
            if t == windows[wn]["s"]:                              # SPY just broke SMA - buffer
                lvl = skip_level(lb, skip, t)
                if not np.isnan(lvl) and P[t] <= lvl:
                    skips += 1
                    marks.append(("skip", t, T[t], "first"))
                elif P[t] > -zb:
                    kind = "first"
                else:
                    blocked += 1
                    marks.append(("blocked", t, T[t], "first"))
            elif DN[t] == nd and P[t] > -zb:
                kind = "turn"
            if kind:
                in_pos, n_entries = True, n_entries + 1
                st = dict(ep=T[t], t0=t, wn=wn, kind=kind)
                marks.append(("entry", t, T[t], kind))
        curve[t] = eq

    if in_pos:
        trades.append(dict(window=st["wn"] + 1, kind=st["kind"], entry=idx[st["t0"]], exit=idx[-1],
                           days=N - 1 - st["t0"], ret=T[-1] / st["ep"] - 1, reason="still open"))
    curve = pd.Series(curve, index=idx)
    wins = [dict(ret=curve.iloc[w["e"]] / curve.iloc[w["s"] - 1] - 1,
                 traded=any(m[0] == "entry" and win_of[m[1]] == wn for m in marks)) for wn, w in enumerate(windows)]
    return dict(curve=curve, trades=pd.DataFrame(trades), marks=marks, wins=wins, skips=skips, blocked=blocked)


REASONS = ["PMO turn up", "PMO upper", "take-profit", "stop loss", "window end", "still open"]

def summarize(res):
    w = pd.DataFrame(res["wins"]).assign(type=[x["type"] for x in windows])
    tr, c, r = res["trades"], res["curve"], np.array([x["ret"] for x in res["wins"]])
    years = (c.index[-1] - c.index[0]).days / 365.25
    out = {"Windows traded": w["traded"].mean(), "Mean window": r.mean(), "Median window": np.median(r),
           "% windows > 0": (r > 0).mean(), "% windows ≥ 10%": (r >= 0.10).mean(),
           "Worst window": r.min(), "Best window": r.max(),
           "Compounded": np.prod(1 + r) - 1, "Compounded ex-best": np.prod(1 + np.sort(r)[:-1]) - 1}
    for k in ["shallow", "flash", "bear"]:
        sub = w[w.type == k]
        out[f"Mean {k} ({len(sub)})"] = sub["ret"].mean() if len(sub) else np.nan
    has = len(tr) > 0
    out.update({"Trades": len(tr), "First entries skipped": res["skips"], "First entries blocked (PMO<0)": res["blocked"],
                "Trade win rate": (tr["ret"] > 0).mean() if has else np.nan,
                "Avg trade": tr["ret"].mean() if has else np.nan,
                "Avg win": tr.loc[tr.ret > 0, "ret"].mean() if has and (tr.ret > 0).any() else np.nan,
                "Avg loss": tr.loc[tr.ret <= 0, "ret"].mean() if has and (tr.ret <= 0).any() else np.nan,
                "Avg days held": tr["days"].mean() if has else np.nan,
                "Sleeve CAGR": c.iloc[-1] ** (1 / years) - 1, "Sleeve max DD": (c / c.cummax() - 1).min()})
    why = tr["reason"].value_counts(normalize=True) if has else pd.Series(dtype=float)
    out.update({f"Exit: {k}": why.get(k, 0.0) for k in REASONS})
    return out


skip_label = lambda sk: "no skip" if sk is None else f"skip {sk[0]}×{sk[1]:g}"
name = lambda nd, nu, lb, sk, zb: (f"PMO turn | in {nd}↓ out {nu}↑ | zero ±{zb:g} | " +
                                   (skip_label(sk) if sk is None else f"lb {lb} | {skip_label(sk)}"))
results = {}
zb_base = ZERO_BUFFERS[0]
for nd in ENTER_DAYS_LIST:
    results[f"Baseline: in {nd}↓ | hold to window end"] = run(nd, zb=zb_base)
    for tp in BASE_TPS:
        results[f"Baseline: in {nd}↓ | TP {pct(tp)}"] = run(nd, tp=tp, zb=zb_base)
for nd in ENTER_DAYS_LIST:
    for nu in EXIT_DAYS_LIST:
        for zb in ZERO_BUFFERS:
            for sk in SKIPS:
                for lb in (LOOKBACKS if sk is not None else [None]):
                    results[name(nd, nu, lb, sk, zb)] = run(nd, nu, lb, sk, zb=zb)
nd0, nu0, lb0, sk0, zb0 = SHOW
show = name(nd0, nu0, lb0 if sk0 is not None else None, sk0, zb0)
if show not in results:                                  # SHOW always runs, even outside the grid
    if sk0 is not None and (sk0[0], lb0) not in REFS:
        REFS[(sk0[0], lb0)] = (lowest_low if sk0[0] == "lowest" else recent_low)(lb0).loc[idx].values
    results[show] = run(nd0, nu0, lb0, sk0, zb=zb0)

# ------------------------- TABLES -------------------------------
types = pd.Series([w["type"] for w in windows]).value_counts()
print(f"{len(windows)} risk-off windows ({BASELINE} SMA{BASE_SMA} ±{pct(BASE_BUFFER)}): {types.get('shallow', 0)} shallow, "
      f"{types.get('flash', 0)} flash, {types.get('bear', 0)} bear | trade {TRADE} | PMO({PMO_LEN1},{PMO_LEN2}) | "
      f"swing lows K={PIVOT_K}{' below 0' if PIVOT_BELOW_ZERO else ''} | stop loss {pct(STOP_LOSS) if STOP_LOSS else 'off'}")
summary = pd.DataFrame({n: summarize(r) for n, r in results.items()}).T
cnt = ["Trades", "First entries skipped", "First entries blocked (PMO<0)"]
display(summary.style.format({c: ("{:.0f}" if c in cnt else "{:.1f}" if c == "Avg days held" else "{:.1%}")
                              for c in summary.columns}, na_rep="–"))

skip_rows = [(sk, lb) for sk in SKIPS for lb in (LOOKBACKS if sk is not None else [None])]
for nd in ENTER_DAYS_LIST:
    for metric in ["Median window", "% windows > 0", "Compounded ex-best"]:
        grid = pd.DataFrame({f"out {nu}↑ | zero ±{zb:g}": [summary.loc[name(nd, nu, lb, sk, zb), metric] for sk, lb in skip_rows]
                             for nu in EXIT_DAYS_LIST for zb in ZERO_BUFFERS},
                            index=[skip_label(sk) + ("" if sk is None else f" | lb {lb}") for sk, lb in skip_rows])
        display(grid.style.format("{:.1%}").background_gradient(cmap="RdYlGn", axis=None)
                .set_caption(f"in {nd}↓ — {metric}"))

res = results[show]
compare = [f"Baseline: in {nd0}↓ | hold to window end"] + [f"Baseline: in {nd0}↓ | TP {pct(tp)}" for tp in BASE_TPS] + \
          [name(nd0, nu0, None, None, zb0), show]
compare = list(dict.fromkeys(n for n in compare if n in results))
W = pd.DataFrame({"Start": [idx[w["s"]].date() for w in windows],
                  "End": [f"{idx[w['e']].date()}{' (open)' if w['open'] else ''}" for w in windows],
                  "Days": [w["e"] - w["s"] for w in windows], "Type": [w["type"] for w in windows],
                  "Max depth": [w["depth"] for w in windows], BASELINE: [B[w["e"]] / B[w["s"]] - 1 for w in windows]})
for n in compare:
    W[n.replace("Baseline: ", "").replace("PMO turn | ", "")] = [x["ret"] for x in results[n]["wins"]]
print(f"Per window — SHOW config: {show}")
display(W.style.format({c: "{:.1%}" for c in W.columns[4:]}))
if len(res["trades"]):
    log = res["trades"].copy()
    log["entry"], log["exit"] = log["entry"].dt.date, log["exit"].dt.date
    print(f"Trade log — {show}")
    display(log.style.format({"ret": "{:.1%}"}))

# ------------------------- PLOT ---------------------------------
type_color = {"shallow": "gold", "flash": "orange", "bear": "red"}
exit_color = {"PMO turn up": "blue", "PMO upper": "magenta", "take-profit": "green", "stop loss": "darkred", "window end": "black",
              "still open": "gray"}
fig, (ax1, ax2, axp, ax3) = plt.subplots(4, 1, figsize=(15, 15), sharex=True,
                                         gridspec_kw={"height_ratios": [3, 3, 2, 1.3]})
ax1.plot(idx, B, color="black", lw=1, label=BASELINE)
ax1.plot(idx, BS, color="tab:blue", lw=1, label=f"SMA {BASE_SMA}")
ax1.plot(idx, BS * (1 - BASE_BUFFER), color="tab:blue", lw=0.7, ls="--", label=f"risk-off line -{pct(BASE_BUFFER)}")
ax1.plot(idx, BS * (1 + BASE_BUFFER), color="tab:blue", lw=0.7, ls=":", label=f"risk-on line +{pct(BASE_BUFFER)}")
ax1.set_yscale("log"); ax1.set_title(f"{BASELINE} — windows shaded by type (yellow shallow / orange flash / red bear)")
ax2.plot(idx, T, color="purple", lw=1, label=f"Trade {TRADE}")
ax2.set_yscale("log")
ax2.set_title(f"{show}  (▲ first, △ turn entry, ✕ red skipped / grey blocked PMO<0, ▼ exit by reason)")
axp.plot(idx, P, color="black", lw=1, label="PMO")
axp.axhline(0, color="gray", lw=0.8)
if zb0: axp.axhspan(-zb0, zb0, color="gray", alpha=0.12, label=f"zero band ±{zb0:g}")
if PMO_UPPER_EXIT is not None:
    axp.axhline(PMO_UPPER_EXIT, color="magenta", lw=0.8, ls="--", label=f"upper exit {PMO_UPPER_EXIT:g}")
if sk0 is not None and (sk0[0], lb0) in REFS:
    axp.plot(idx, REFS[(sk0[0], lb0)] * sk0[1], color="crimson", lw=1, ls="--",
             label=f"skip line: {sk0[0]} swing low × {sk0[1]:g} ({lb0} bars)")
pv_in = [k for k in range(len(piv_i)) if pmo.index[piv_i[k]] >= idx[0]]
axp.scatter(pmo.index[piv_i[pv_in]], piv_v[pv_in], s=10, color="crimson", zorder=3, label="swing lows")
axp.set_title("PMO with swing lows, skip line, entries (▲) and exits (▼)")
for w in windows:
    for ax in (ax1, ax2, axp, ax3):
        ax.axvspan(idx[w["s"]], idx[w["e"]], color=type_color[w["type"]], alpha=0.12, lw=0)
entry_t = None
for kind, t, p, why in res["marks"]:
    if kind == "entry":
        entry_t = t
        style = dict(color="purple") if why == "first" else dict(color="white", edgecolor="purple")
        for ax, y in ((ax2, p), (ax1, B[t]), (axp, P[t])):
            ax.scatter(idx[t], y, marker="^", s=34, zorder=4, **style)
    elif kind in ("skip", "blocked"):
        for ax, y in ((ax2, p), (axp, P[t])):
            ax.scatter(idx[t], y, marker="x", s=40, zorder=4, color="crimson" if kind == "skip" else "gray")
    else:
        for ax, y in ((ax2, p), (ax1, B[t]), (axp, P[t])):
            ax.scatter(idx[t], y, marker="v", s=34, zorder=4, color=exit_color[why])
        ax2.axvspan(idx[entry_t], idx[t], color="gray", alpha=0.18, lw=0)
for why, c in exit_color.items():
    ax2.scatter([], [], marker="v", color=c, label=f"exit: {why}")
ax3.plot(res["curve"].index, res["curve"].values, color="tab:green", lw=1.2, label="Sleeve equity (cash when flat)")
ax3.set_yscale("log")
for ax in (ax1, ax2, axp, ax3): ax.grid(alpha=0.3); ax.legend(loc="upper left", fontsize=8, ncol=2)
plt.tight_layout(); plt.show()
```

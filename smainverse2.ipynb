```python
# ==============================================================
# Risk-OFF KST + Williams VIX Fix Tester  (standalone Jupyter cell)
# --------------------------------------------------------------
# Window      : BASELINE closes below SMA*(1-BASE_BUFFER) ... until it closes above SMA*(1+BASE_BUFFER)
#               No trade on the window-open day (logged in the "Window opens" table for review).
# KST (Pring) : sum of w_i * SMA(ROC(close, r_i), s_i), ROC 10/15/20/30, SMA 10/10/10/15, weights 1..4,
#               signal = SMA9(KST)   (matches TradingView "Know sure thing")
# WVF         : (highest close 22 - low) / highest close 22 * 100   (TheeHoganator "Lows" style)
#               green bar = WVF >= upper Bollinger (SMA20 + 2*stdev20, population stdev) OR WVF >= 0.85 * highest
#               WVF of 50 bars (WVF_GREEN = "band" uses the Bollinger condition only)
#               FIRE  = WVF_RISING_DAYS green bars in a row, each WVF higher than the last,
#                       or WVF_GREEN_DAYS green bars in a row
#               FADE  = green yesterday, grey today
# Entry       : inside a window, while flat: KST has just completed ENTER_DOWNTICKS falling closes in a row,
#               KST > -zero buffer, KST <= KST_CAP, and BASELINE is at/below its SMA. Buy TRADE at that close.
# Exits (first one wins; all at the close unless noted):
#   duped      : KST turns up (UPTICKS rising closes) and the turn started at/above +zero buffer
#                (momentum never really broke — we were duped)
#   bottom     : WVF has fired since entry ("armed"), then
#                  "kst"    -> KST turns up (UPTICKS rising closes) from a trough below +zero buffer
#                  "fade"   -> WVF fades while KST is below +zero buffer
#                  "either" -> whichever comes first
#   special    : KST went below zero, WVF never fired, and KST came back above +zero buffer -> from then on sell
#                if KST falls back to its entry value ("kst") or TRADE trades back to its entry price
#                ("price", intraday stop)
#   window end : sold at the close the day BASELINE turns risk-on
#   optional hard stop loss (intraday)
# ==============================================================
import numpy as np
import pandas as pd
import yfinance as yf
import matplotlib.pyplot as plt
from IPython.display import display

# ------------------------- INPUTS -------------------------------
BASELINE       = "SPY"
BASE_SMA       = 200
BASE_BUFFER    = 0.04
TRADE          = {"SPY": -3.0}        # traded inverse (synthetic SPXU); real ETF: {"SPXU": 1}
TRADE_COST     = 0.0                  # annual drag on TRADE (e.g. 0.03)
STOP_LOSS      = None                 # e.g. 0.15; None = off

# KST (TradingView defaults)
ROC_LENS       = (10, 15, 20, 30)
SMA_LENS       = (10, 10, 10, 15)
KST_SIGNAL     = 9
KST_CAP        = 50                   # no entries while KST is above this
ENTER_DOWNTICKS_LIST = [1, 2, 3]
UPTICKS_LIST   = [1, 2]               # rising KST closes that count as a "turn up" (duped + bottom exits)
ZERO_BUFFERS   = [0.0, 2.0]           # KST units: entries need KST > -b; a turn from >= +b is "duped"

# Williams VIX Fix (TradingView defaults)
WVF_PD, WVF_BBL, WVF_MULT, WVF_LB, WVF_PH = 22, 20, 2.0, 50, 0.85
WVF_GREEN      = "script"             # "script" = band OR 85% of 50-bar high (as TradingView); "band" = band only
WVF_RISING_DAYS = 4                   # fire: this many green bars in a row, each higher
WVF_GREEN_DAYS  = 6                   # fire: or this many green bars in a row

BOTTOM_EXITS   = ["kst"]
SPECIAL_STOPS  = ["kst", "price"]     # rare-case stop reference: entry KST or entry price

BASE_TPS       = [0.04]         # baselines on the same entries: fixed TP + window end

SHALLOW_DEPTH, FLASH_DAYS = 0.08, 90  # hindsight window tags (analysis only)
CASH_RATE      = 0.03
START, END     = "1994-01-01", "2026-10-25"
SHOW           = (2, 3, "kst", 0.0, "price")   # (ENTER_DOWNTICKS, UPTICKS, BOTTOM_EXIT, ZERO_BUFFER, SPECIAL_STOP)
TRADING_DAYS   = 252
PLOT_FULL      = True
PLOT_WINDOWS   = True
WINDOW_PAD     = (60, 40)             # bars before / after each window in the per-window charts
PLOT_WINDOW_TYPES = None              # e.g. ["bear", "flash"]; None = all
# ----------------------------------------------------------------

pct = lambda x: f"{x * 100:g}%"
tickers = sorted({BASELINE, *TRADE})
warmup = pd.Timestamp(START) - pd.Timedelta(days=int(max(BASE_SMA, max(ROC_LENS) + max(SMA_LENS) + KST_SIGNAL) * 1.6) + 60)
raw = yf.download(tickers, start=warmup, end=END, auto_adjust=True, progress=False)
if not isinstance(raw.columns, pd.MultiIndex):
    raw.columns = pd.MultiIndex.from_product([raw.columns, tickers])
raw = raw[["Open", "High", "Low", "Close"]].dropna(how="any")
px = raw["Close"]
rets = px.pct_change()

def synth(weights, cost=0.0):
    w = pd.Series(weights, dtype=float)
    r = rets[w.index].mul(w, axis=1).clip(lower=-w.abs(), axis=1).sum(axis=1)
    r = (r - cost / TRADING_DAYS).clip(lower=-1.0); r.iloc[0] = 0.0
    return (1 + r).cumprod()

def synth_ohl(weights, close):
    w = pd.Series(weights, dtype=float)
    prev = px[w.index].shift(1)
    move = lambda f: raw[f][w.index] / prev - 1
    o = (move("Open") * w).sum(axis=1)
    hi = sum(wt * move("High" if wt > 0 else "Low")[t] for t, wt in w.items())
    lo = sum(wt * move("Low" if wt > 0 else "High")[t] for t, wt in w.items())
    base = close.shift(1)
    return base * (1 + o).clip(lower=0), base * (1 + hi), base * (1 + lo).clip(lower=0)

trd = synth(TRADE, TRADE_COST)
trd_o, trd_h, trd_l = synth_ohl(TRADE, trd)

b, b_low = px[BASELINE], raw["Low"][BASELINE]
b_sma = b.rolling(BASE_SMA).mean()
g, cur = np.full(len(b), np.nan), None
for k_, (p, s) in enumerate(zip(b.values, b_sma.values)):
    if np.isnan(s): continue
    if cur is None: cur = p > s
    elif cur and p < s * (1 - BASE_BUFFER): cur = False
    elif not cur and p > s * (1 + BASE_BUFFER): cur = True
    g[k_] = cur
gate = pd.Series(g, index=b.index)

# KST
roc = lambda n: 100 * (b / b.shift(n) - 1)
kst = sum(w * roc(r).rolling(s).mean() for w, r, s in zip((1, 2, 3, 4), ROC_LENS, SMA_LENS))
kst_sig = kst.rolling(KST_SIGNAL).mean()

# Williams VIX Fix
hc = b.rolling(WVF_PD).max()
wvf = (hc - b_low) / hc * 100
wvf_mid = wvf.rolling(WVF_BBL).mean()
wvf_up = wvf_mid + WVF_MULT * wvf.rolling(WVF_BBL).std(ddof=0)
wvf_rh = wvf.rolling(WVF_LB).max() * WVF_PH
green = (wvf >= wvf_up) | ((wvf >= wvf_rh) if WVF_GREEN == "script" else False)

idx = px.index[(px.index >= pd.Timestamp(START)) & gate.notna() & kst_sig.notna() & wvf_up.notna()]
B, BS = b.loc[idx].values, b_sma.loc[idx].values
T, TO, TH, TL = (x.loc[idx].values for x in (trd, trd_o, trd_h, trd_l))
K, KS = kst.loc[idx].values, kst_sig.loc[idx].values
WV, WU, WR = wvf.loc[idx].values, wvf_up.loc[idx].values, wvf_rh.loc[idx].values
GR = green.loc[idx].values.astype(bool)
ON = gate.loc[idx].astype(bool).values
N = len(idx)

# streaks
DN, UP, GRUN, RRUN = (np.zeros(N, int) for _ in range(4))
for t in range(1, N):
    DN[t] = DN[t-1] + 1 if K[t] < K[t-1] else 0
    UP[t] = UP[t-1] + 1 if K[t] > K[t-1] else 0
    GRUN[t] = GRUN[t-1] + 1 if GR[t] else 0
    RRUN[t] = (RRUN[t-1] + 1 if GR[t-1] and WV[t] > WV[t-1] else 1) if GR[t] else 0
FIRE = (RRUN >= WVF_RISING_DAYS) | (GRUN >= WVF_GREEN_DAYS)
FADE = np.r_[False, GR[:-1] & ~GR[1:]]

windows, i = [], 1
while i < N:
    if ON[i-1] and not ON[i]:
        j = i
        while j < N and not ON[j]: j += 1
        e = min(j, N - 1)
        depth = (B[i:e+1] / BS[i:e+1] - 1).min()
        kind = "shallow" if depth > -SHALLOW_DEPTH else ("flash" if e - i <= FLASH_DAYS else "bear")
        windows.append(dict(s=i, e=e, open=j >= N, depth=depth, type=kind))
        i = j
    i += 1
if not windows:
    raise ValueError("No risk-off windows in this date range.")
win_of = np.full(N, -1)
for wn, w in enumerate(windows):
    win_of[w["s"]:w["e"] + 1] = wn


def run(nd, nu=1, bottom="either", zb=0.0, special="kst", tp=None, baseline=False):
    """baseline=True: same entries, exits only by TP / window end / stop loss."""
    cash_d = (1 + CASH_RATE) ** (1 / TRADING_DAYS) - 1
    eq, curve, trades, marks = 1.0, np.ones(N), [], []
    in_pos, st, blocked = False, {}, dict(above_sma=0, cap=0)

    for t in range(1, N):
        wn = win_of[t]
        exited_today = False
        if in_pos:
            ep, hit = st["ep"], None
            if STOP_LOSS:
                L = ep * (1 - STOP_LOSS)
                if TO[t] <= L: hit = (TO[t], "stop loss")
                elif TL[t] <= L: hit = (L, "stop loss")
            if hit is None and not baseline and st["special"] and special == "price":
                if TO[t] <= ep: hit = (TO[t], "special stop")
                elif TL[t] <= ep: hit = (ep, "special stop")
            if hit is None and tp is not None:
                L = ep * (1 + tp)
                if TO[t] >= L: hit = (TO[t], "take-profit")
                elif TH[t] >= L: hit = (L, "take-profit")
            if FIRE[t]: st["armed"] = True
            if K[t] < 0: st["went_below"] = True
            if not baseline and hit is None:
                turn = UP[t] == nu
                trough = K[t - UP[t]] if UP[t] else K[t]
                if turn and trough >= zb:
                    hit = (T[t], "duped")
                elif st["armed"] and turn and trough < zb and bottom in ("kst", "either"):
                    hit = (T[t], "bottom: KST turn")
                elif st["armed"] and FADE[t] and K[t] < zb and bottom in ("fade", "either"):
                    hit = (T[t], "bottom: WVF fade")
                elif st["special"] and special == "kst" and K[t] <= st["ek"]:
                    hit = (T[t], "special stop")
            if hit is None and ON[t]: hit = (T[t], "window end")
            if hit:
                fill, why = hit
                eq *= fill / T[t-1]
                trades.append(dict(window=st["wn"] + 1, entry=idx[st["t0"]], exit=idx[t], days=t - st["t0"],
                                   ret=fill / ep - 1, reason=why, entry_KST=st["ek"], exit_KST=K[t],
                                   min_KST=st["mk"], WVF_armed=st["armed"]))
                marks.append(("exit", t, fill, why))
                in_pos, exited_today = False, True
            else:
                eq *= T[t] / T[t-1]
                st["mk"] = min(st["mk"], K[t])
                if not st["special"] and st["went_below"] and not st["armed"] and K[t] >= zb and K[t-1] < zb:
                    st["special"] = True
                    marks.append(("special", t, T[t], None))
        else:
            eq *= 1 + cash_d

        if not in_pos and not exited_today and wn >= 0 and not ON[t] and t != windows[wn]["s"]:
            if DN[t] == nd and K[t] > -zb:
                if B[t] > BS[t]:
                    blocked["above_sma"] += 1
                elif K[t] > KST_CAP:
                    blocked["cap"] += 1
                else:
                    in_pos = True
                    st = dict(ep=T[t], t0=t, wn=wn, ek=K[t], mk=K[t], armed=False, went_below=K[t] < 0,
                              special=False)
                    marks.append(("entry", t, T[t], None))
        curve[t] = eq

    if in_pos:
        trades.append(dict(window=st["wn"] + 1, entry=idx[st["t0"]], exit=idx[-1], days=N - 1 - st["t0"],
                           ret=T[-1] / st["ep"] - 1, reason="still open", entry_KST=st["ek"], exit_KST=K[-1],
                           min_KST=st["mk"], WVF_armed=st["armed"]))
    curve = pd.Series(curve, index=idx)
    wins = [dict(ret=curve.iloc[w["e"]] / curve.iloc[w["s"] - 1] - 1,
                 traded=any(m[0] == "entry" and win_of[m[1]] == wn for m in marks)) for wn, w in enumerate(windows)]
    return dict(curve=curve, trades=pd.DataFrame(trades), marks=marks, wins=wins, blocked=blocked)


REASONS = ["duped", "bottom: KST turn", "bottom: WVF fade", "special stop", "take-profit", "stop loss",
           "window end", "still open"]

def summarize(res):
    w = pd.DataFrame(res["wins"]).assign(type=[x["type"] for x in windows])
    tr, c, r = res["trades"], res["curve"], np.array([x["ret"] for x in res["wins"]])
    years = (c.index[-1] - c.index[0]).days / 365.25
    out = {"Windows traded": w["traded"].mean(), "Mean window": r.mean(), "Median window": np.median(r),
           "% windows > 0": (r > 0).mean(), "% windows ≥ 10%": (r >= 0.10).mean(),
           "Worst window": r.min(), "Best window": r.max(),
           "Compounded": np.prod(1 + r) - 1, "Compounded ex-best": np.prod(1 + np.sort(r)[:-1]) - 1}
    for k in ["shallow", "flash", "bear"]:
        sub = w[w.type == k]
        out[f"Mean {k} ({len(sub)})"] = sub["ret"].mean() if len(sub) else np.nan
    has = len(tr) > 0
    out.update({"Trades": len(tr),
                "Blocked: SPY > SMA": res["blocked"]["above_sma"], "Blocked: KST > cap": res["blocked"]["cap"],
                "Trade win rate": (tr["ret"] > 0).mean() if has else np.nan,
                "Avg trade": tr["ret"].mean() if has else np.nan,
                "Avg win": tr.loc[tr.ret > 0, "ret"].mean() if has and (tr.ret > 0).any() else np.nan,
                "Avg loss": tr.loc[tr.ret <= 0, "ret"].mean() if has and (tr.ret <= 0).any() else np.nan,
                "% trades WVF armed": tr["WVF_armed"].mean() if has else np.nan,
                "Avg days held": tr["days"].mean() if has else np.nan,
                "Sleeve CAGR": c.iloc[-1] ** (1 / years) - 1, "Sleeve max DD": (c / c.cummax() - 1).min()})
    why = tr["reason"].value_counts(normalize=True) if has else pd.Series(dtype=float)
    out.update({f"Exit: {k}": why.get(k, 0.0) for k in REASONS})
    return out


name = lambda nd, nu, bx, zb, sp: f"KST in {nd}↓ | out {nu}↑ | bottom {bx} | zero ±{zb:g} | special {sp}"
results = {}
for nd in ENTER_DOWNTICKS_LIST:
    results[f"Baseline: in {nd}↓ | hold to window end"] = run(nd, baseline=True)
    for tp in BASE_TPS:
        results[f"Baseline: in {nd}↓ | TP {pct(tp)}"] = run(nd, tp=tp, baseline=True)
for nd in ENTER_DOWNTICKS_LIST:
    for nu in UPTICKS_LIST:
        for bx in BOTTOM_EXITS:
            for zb in ZERO_BUFFERS:
                for sp in SPECIAL_STOPS:
                    results[name(nd, nu, bx, zb, sp)] = run(nd, nu, bx, zb, sp)
nd0, nu0, bx0, zb0, sp0 = SHOW
show = name(*SHOW)
if show not in results:                                     # SHOW always runs, even outside the grid
    results[show] = run(nd0, nu0, bx0, zb0, sp0)

# ------------------------- TABLES -------------------------------
types = pd.Series([w["type"] for w in windows]).value_counts()
print(f"{len(windows)} risk-off windows ({BASELINE} SMA{BASE_SMA} ±{pct(BASE_BUFFER)}): {types.get('shallow', 0)} shallow, "
      f"{types.get('flash', 0)} flash, {types.get('bear', 0)} bear | trade {TRADE} | KST cap {KST_CAP} | "
      f"WVF fire: {WVF_RISING_DAYS} rising green or {WVF_GREEN_DAYS} green ({WVF_GREEN}) | "
      f"stop loss {pct(STOP_LOSS) if STOP_LOSS else 'off'}")
print(f"KST range over the test period: {np.nanmin(K):.1f} to {np.nanmax(K):.1f} "
      f"(inside windows: {np.nanmin(K[win_of >= 0]):.1f} to {np.nanmax(K[win_of >= 0]):.1f})")
summary = pd.DataFrame({n: summarize(r) for n, r in results.items()}).T
cnt = ["Trades", "Blocked: SPY > SMA", "Blocked: KST > cap"]
display(summary.style.format({c: ("{:.0f}" if c in cnt else "{:.1f}" if c == "Avg days held" else "{:.1%}")
                              for c in summary.columns}, na_rep="–"))

for nd in ENTER_DOWNTICKS_LIST:
    for metric in ["Median window", "% windows > 0", "Compounded ex-best"]:
        grid = pd.DataFrame({f"out {nu}↑ | zero ±{zb:g}": [summary.loc[name(nd, nu, bx, zb, sp), metric]
                                                           for bx in BOTTOM_EXITS for sp in SPECIAL_STOPS]
                             for nu in UPTICKS_LIST for zb in ZERO_BUFFERS},
                            index=[f"bottom {bx} | special {sp}" for bx in BOTTOM_EXITS for sp in SPECIAL_STOPS])
        display(grid.style.format("{:.1%}").background_gradient(cmap="RdYlGn", axis=None)
                .set_caption(f"in {nd}↓ — {metric}"))

# window opens (not traded) — what would have happened
opens = pd.DataFrame([{
    "Window": wn + 1, "Open date": idx[w["s"]].date(), "Type": w["type"],
    "KST at open": K[w["s"]], "KST falling?": K[w["s"]] < K[w["s"] - 1],
    "WVF green?": GR[w["s"]],
    "TRADE +10d": T[min(w["s"] + 10, w["e"])] / T[w["s"]] - 1,
    "TRADE +20d": T[min(w["s"] + 20, w["e"])] / T[w["s"]] - 1,
    "TRADE best within window": T[w["s"]:w["e"] + 1].max() / T[w["s"]] - 1,
    "TRADE at window end": T[w["e"]] / T[w["s"]] - 1} for wn, w in enumerate(windows)])
print("Window opens (not traded) — inverse return had you bought at the open")
display(opens.style.format({c: "{:.1%}" for c in opens.columns if c.startswith("TRADE")} | {"KST at open": "{:.1f}"}))

res = results[show]
compare = [f"Baseline: in {nd0}↓ | hold to window end"] + [f"Baseline: in {nd0}↓ | TP {pct(tp)}" for tp in BASE_TPS] + [show]
W = pd.DataFrame({"Start": [idx[w["s"]].date() for w in windows],
                  "End": [f"{idx[w['e']].date()}{' (open)' if w['open'] else ''}" for w in windows],
                  "Days": [w["e"] - w["s"] for w in windows], "Type": [w["type"] for w in windows],
                  "Max depth": [w["depth"] for w in windows], BASELINE: [B[w["e"]] / B[w["s"]] - 1 for w in windows]})
for n in compare:
    W[n.replace("Baseline: ", "")] = [x["ret"] for x in results[n]["wins"]]
print(f"Per window — SHOW config: {show}")
display(W.style.format({c: "{:.1%}" for c in W.columns[4:]}))
if len(res["trades"]):
    log = res["trades"].copy()
    log["entry"], log["exit"] = log["entry"].dt.date, log["exit"].dt.date
    print(f"Trade log — {show}")
    display(log.style.format({"ret": "{:.1%}", "entry_KST": "{:.1f}", "exit_KST": "{:.1f}", "min_KST": "{:.1f}"}))

# ------------------------- PLOT ---------------------------------
type_color = {"shallow": "gold", "flash": "orange", "bear": "red"}
exit_color = {"duped": "orange", "bottom: KST turn": "blue", "bottom: WVF fade": "deepskyblue",
              "special stop": "brown", "take-profit": "green", "stop loss": "darkred", "window end": "black",
              "still open": "gray"}
curve = res["curve"].values


def draw(a, z, title, rebase=False, figsize=(15, 17)):
    sl = slice(a, z + 1)
    x = idx[sl]
    fig, (ax1, ax2, axk, axw, ax3) = plt.subplots(5, 1, figsize=figsize, sharex=True,
                                                  gridspec_kw={"height_ratios": [3, 3, 2, 1.6, 1.3]})
    fig.suptitle(title, fontsize=12, y=0.995)
    ax1.plot(x, B[sl], color="black", lw=1, label=BASELINE)
    ax1.plot(x, BS[sl], color="tab:blue", lw=1, label=f"SMA {BASE_SMA}")
    ax1.plot(x, BS[sl] * (1 - BASE_BUFFER), color="tab:blue", lw=0.7, ls="--", label=f"risk-off -{pct(BASE_BUFFER)}")
    ax1.plot(x, BS[sl] * (1 + BASE_BUFFER), color="tab:blue", lw=0.7, ls=":", label=f"risk-on +{pct(BASE_BUFFER)}")
    ax1.set_yscale("log"); ax1.set_title(f"{BASELINE} — windows shaded by type (yellow shallow / orange flash / red bear)")
    ax2.plot(x, T[sl], color="purple", lw=1, label=f"Trade {TRADE}")
    ax2.set_yscale("log"); ax2.set_title(f"{show}  (▲ entry, ✕ window open (not traded), ◆ special armed, ▼ exit)")
    axk.plot(x, K[sl], color="#009688", lw=1.1, label="KST")
    axk.plot(x, KS[sl], color="#F44336", lw=0.8, label=f"signal SMA{KST_SIGNAL}")
    axk.axhline(0, color="gray", lw=0.8)
    if zb0: axk.axhspan(-zb0, zb0, color="gray", alpha=0.12, label=f"zero band ±{zb0:g}")
    axk.axhline(KST_CAP, color="magenta", lw=0.8, ls="--", label=f"entry cap {KST_CAP}")
    axk.set_title("KST")
    axw.bar(x, WV[sl], color=np.where(GR[sl], "#00CC00", "#9E9E9E"), width=1.0, label="WVF (green = above band / range)")
    axw.plot(x, WU[sl], color="aqua", lw=0.9, label="upper Bollinger")
    if WVF_GREEN == "script": axw.plot(x, WR[sl], color="orange", lw=0.8, ls=":", label=f"{WVF_PH:g} × {WVF_LB}-bar high")
    fx = np.nonzero(FIRE[sl])[0]
    axw.scatter(x[fx], WV[sl][fx] * 1.05, marker="*", s=30, color="red", zorder=4, label="FIRE")
    axw.set_title("Williams VIX Fix")
    for w in windows:
        if w["e"] < a or w["s"] > z: continue
        for ax in (ax1, ax2, axk, axw, ax3):
            ax.axvspan(idx[max(w["s"], a)], idx[min(w["e"], z)], color=type_color[w["type"]], alpha=0.12, lw=0)
        if a <= w["s"] <= z:
            ax2.scatter(idx[w["s"]], T[w["s"]], marker="x", s=46, color="gray", zorder=4)
            axk.scatter(idx[w["s"]], K[w["s"]], marker="x", s=46, color="gray", zorder=4)
    entry_t = None
    for kind, t, pr, why in res["marks"]:
        if kind == "entry": entry_t = t
        if not a <= t <= z: continue
        if kind == "entry":
            for ax, y in ((ax2, pr), (ax1, B[t]), (axk, K[t])):
                ax.scatter(idx[t], y, marker="^", s=40, zorder=4, color="purple")
        elif kind == "special":
            for ax, y in ((ax2, pr), (axk, K[t])):
                ax.scatter(idx[t], y, marker="D", s=30, zorder=4, color="brown")
        else:
            for ax, y in ((ax2, pr), (ax1, B[t]), (axk, K[t])):
                ax.scatter(idx[t], y, marker="v", s=40, zorder=4, color=exit_color[why])
            if entry_t is not None:
                ax2.axvspan(idx[max(entry_t, a)], idx[t], color="gray", alpha=0.18, lw=0)
    for why, c in exit_color.items():
        ax2.scatter([], [], marker="v", color=c, label=f"exit: {why}")
    eqv = curve[sl] / curve[a] if rebase else curve[sl]
    ax3.plot(x, eqv, color="tab:green", lw=1.2,
             label="Sleeve equity, rebased to 1 at chart start" if rebase else "Sleeve equity (cash when flat)")
    if rebase: ax3.axhline(1, color="gray", lw=0.8)
    else: ax3.set_yscale("log")
    for ax in (ax1, ax2, axk, axw, ax3): ax.grid(alpha=0.3); ax.legend(loc="upper left", fontsize=8, ncol=2)
    plt.tight_layout(); plt.show()


if PLOT_FULL:
    draw(0, N - 1, f"Full history — {show}")
if PLOT_WINDOWS:
    pad_a, pad_z = WINDOW_PAD
    for wn, w in enumerate(windows):
        if PLOT_WINDOW_TYPES and w["type"] not in PLOT_WINDOW_TYPES: continue
        tr_w = res["trades"][res["trades"]["window"] == wn + 1] if len(res["trades"]) else res["trades"]
        title = (f"Window {wn + 1}/{len(windows)}: {idx[w['s']].date()} → {idx[w['e']].date()}"
                 f"{' (open)' if w['open'] else ''} | {w['type']} | {w['e'] - w['s']} days | "
                 f"max depth {w['depth']:.1%} | {BASELINE} {B[w['e']] / B[w['s']] - 1:+.1%} | "
                 f"window return {res['wins'][wn]['ret']:+.1%} | {len(tr_w)} trades")
        draw(max(0, w["s"] - pad_a), min(N - 1, w["e"] + pad_z), title, rebase=True, figsize=(15, 14))
```
